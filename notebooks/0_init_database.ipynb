# Initialize the database containing museums and cities data

This notebook loads museums data from wikipedia, and cities data from GeoNames

## 1. Import Required Libraries and Modules
We start by importing the necessary libraries and the custom modules from the museums_db package.

```python
import sys
import os

import museums_db
import pandas as pd

from pprint import pprint

# Configure database connection based on environment
if os.environ.get('JUPYTER_CONTAINER'):
    # Running inside Docker container
    os.environ['DB_HOST'] = 'db'
else:
    # Running locally
    os.environ['DB_HOST'] = 'localhost'
    
```

## 2. Initialize Database
Set up the MySQL database and tables 'museums' and 'cities'.

```python
# Initialize the database
museums_db.init_db()
```

## 3. Retrieve and Store Museum Data
Fetch the list of museums with more than 2 million annual visitors from Wikipedia and store them in the database.

```python
df_museums = museums_db.get_most_visited_museums()
print(f"Loaded {len(df_museums)} museums")
df_museums.head()

museums_db.insert_museums_data(df_museums)
```

## 4. Retrieve and Store City Population Data
For each city, retrieve its population and store it in the database.

```python
df_cities_countries = museums_db.extract_cities_and_countries(df_museums)
museums_db.get_cities_population(df_cities_countries)
df_cities_countries.head()

museums_db.insert_cities_data(df_cities_countries)
```

## 5. Show data stored in database


```python
museum_city_data: list[dict] = museums_db.get_museum_city_data()
df_select_data = pd.json_normalize(museum_city_data)

print(f"Loaded {len(df_select_data)} museum-city entries")

df_select_data.head()
```
